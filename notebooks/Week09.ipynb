```python
import sqlite3 #Load SQLite support for database operations.
import pandas as pd #Import pandas as pd for working with tabular data.
conn = sqlite3.connect('my_data.db') #Open the local database, creating the file if it does not exist.
cursor = conn.cursor() #Create a cursor for running SQL commands.
```

Task 1  Sales

```python
# Show the sale date, product, and price for each sales record.
# Load the sales CSV and replace the sales table in the database.
pd.read_csv('../data/csv/sales.csv').to_sql('sales', conn, if_exists = 'replace', index = False) #Read the sales CSV and write its rows to the database.

# Return only the sale date, product, and price columns.
pd.read_sql_query('''
SELECT sale_date, product, price FROM sales
''', conn)
```

Task 02

```python
# Find Chicago employees with more than 10 years of experience using WHERE and AND.
# Load the employee CSV and replace the employees table in the database.
pd.read_csv('../data/csv/hr_employees.csv').to_sql('employees', conn, if_exists = 'replace', index = False) #Read the employee CSV and store its rows in the database.

# Keep employees whose city is Chicago and whose experience exceeds 10 years.
pd.read_sql_query('''
SELECT name, city, department, salary
FROM employees
WHERE city = 'Chicago' and years_experience > 10
''', conn) #Return the matching employees' names, cities, departments, and salaries.
```

Task 03 Real Eastate

```python
# List the three properties with the greatest square footage using ORDER BY and LIMIT.
# Load the listing CSV and replace the listings table in the database.
pd.read_csv('../data/csv/realestate_listings.csv').to_sql('listings', conn, if_exists = 'replace', index = False) #Read the real estate CSV and store its rows in the database.

# Sort by square footage from largest to smallest and return the three highest-ranked listings.
pd.read_sql_query('''
SELECT listing_id, neighborhood, sqft, list_price
FROM listings
ORDER BY sqft DESC
LIMIT 3
''', conn) #Return the selected property details for the top three listings.
```

Task 04 Marketing

```python
# Summarize weekly impressions and ad spend to compare campaign activity.
# Load the marketing campaign CSV and replace the campaigns table.
pd.read_csv('../data/csv/marketing_campaigns.csv').to_sql('campaigns', conn, if_exists = 'replace', index = False)

# Group records by week and calculate total impressions and spend for each group.
pd.read_sql_query('''
SELECT week, SUM(impressions) AS total_impressions, SUM(spend) AS total_spend
FROM campaigns
GROUP BY week
''', conn)
```

Task 05 Reastraunt

```python
# Find order dates when combined restaurant sales exceed $400.
# Load the restaurant orders CSV and replace the orders table.
pd.read_csv('../data/csv/restaurant_orders.csv').to_sql('orders', conn, if_exists = 'replace', index = False)

# Add order totals for each date and keep dates with revenue above $400.
pd.read_sql_query('''
SELECT order_date, SUM(order_total) AS total_revenue
FROM orders
GROUP BY order_date
HAVING SUM(order_total) > 400
''', conn)
```

Task 06  Healthcare

```python
# Calculate total visits, average patient age, and maximum wait time across the clinic dataset.
# Load the clinic visits CSV and replace the visits table.
pd.read_csv('../data/csv/clinic_visits.csv').to_sql('visits', conn, if_exists = 'replace', index = False)

# Summarize all records with a count, rounded average, and maximum value.
pd.read_sql_query('''
SELECT doctor,
       ROUND(AVG(patient_age), 1) AS avg_patient_age,
       MAX(wait_minutes) AS longest_wait
FROM visits
GROUP BY doctor
''', conn)
```

Task 7  Banking

```python
# Calculate each branch's total transaction amount by joining customer and transaction data.
pd.read_csv('../data/csv/bank_customers.csv').to_sql('customers', conn, if_exists = 'replace', index = False)
# Load the transaction CSV and replace the transactions table.
pd.read_csv('../data/csv/bank_transactions.csv').to_sql('transactions', conn, if_exists = 'replace', index = False)

# Match transactions to customers, then group transaction amounts by branch.
pd.read_sql_query('''
SELECT c.branch, SUM(t.amount) AS total_transactions
FROM transactions AS t
JOIN customers AS c ON t.customer_id = c.customer_id
GROUP BY c.branch
''', conn)
```
