```python
import sqlite3 #Load SQLite support for database operations.
import pandas as pd #Import pandas as pd for working with tabular data.

#Open the local database, creating the file if it does not exist.
conn = sqlite3.connect('my_data.db') 

 #Create a cursor for running SQL commands.
cursor = conn.cursor()
```

Task 02: Which Seniors Use AI Most for Assignments?
</h3>

<p>
<strong>Question:</strong>
Which senior students reported using AI on assignments?
Display their <code>Major</code>, <code>AI_Tool</code>,
and <code>AI_Hours_Per_Week</code> for students who
answered <code>Yes</code> to <code>Used_AI_Assignment</code>.
Arrange the results so that the heaviest AI users appear first.
</p>

</div>

```python
# Display seniors who used AI for assignments, sorted by weekly usage in descending order
pd.read_sql_query("""
SELECT Major, AI_Tool, AI_Hours_Per_Week
FROM course_evaluation
WHERE Academic_Standing = 'Senior'
AND Used_AI_Assignment = 'Yes'
ORDER BY AI_Hours_Per_Week DESC
""", conn)
```

Task 03: Which AI Tools Are Most Commonly Reported?
</h3>

<p>
<strong>Question:</strong>
Which <code>AI_Tool</code> entries appear most frequently
in the survey? Display the number of students reporting
each tool entry, arranged from most to least common.
Identify the most frequently reported entry.
</p>

</div>

```python
# Count AI tool entries and sort by frequency in descending order
pd.read_sql_query("""
SELECT AI_Tool, COUNT(*) AS Tool_Count
FROM course_evaluation
GROUP BY AI_Tool
ORDER BY Tool_Count DESC
""", conn)
```

Task 04: Does AI Proficiency Relate to Learning Impact?
</h3>

<p>
<strong>Question:</strong>
How does the average <code>AI_Learning_Impact</code>
differ across <code>AI_Proficiency</code> levels?
For each proficiency level, report the number of students
and the average learning impact rounded to two decimal places.
Identify the proficiency level with the highest average
learning impact.
</p>

</div>

```python
# Count AI tool entries and sort by frequency in descending order
pd.read_sql_query("""
SELECT AI_Proficiency,
        COUNT(*) AS Student_Count,
        ROUND(AVG(AI_Learning_Impact), 2) AS AVG_Learning_Impact
FROM course_evaluation
GROUP BY AI_Proficiency
ORDER BY AVG_Learning_Impact DESC
""", conn)
```

Task 05: Which Frequency Groups Spend More Time Using AI?
</h3>

<p>
<strong>Question:</strong>
Which <code>AI_Usage_Frequency</code> groups average
more than one hour per week using AI?
Display each qualifying group and its average
<code>AI_Hours_Per_Week</code>.
Include only groups whose average exceeds one hour.
</p>

</div>

```python
# Display AI usage frequency groups averaging more than one hour per week
pd.read_sql_query("""
SELECT AI_Usage_Frequency,
       AVG(AI_Hours_Per_Week) AS Avg_Hours_Per_Week
FROM course_evaluation
GROUP BY AI_Usage_Frequency
HAVING AVG(AI_Hours_Per_Week) > 1
ORDER BY Avg_Hours_Per_Week DESC
""", conn)
```
